```python
import pandas as pd
import numpy as np
```

```python
risk_df = pd.read_csv("../data/raw/thyroid_cancer_risk_data.csv")

recurrence_df = pd.read_csv("../data/raw/recurrence_dataset.csv")
```

```python
#remove useless columns
risk_df = risk_df.drop("Patient_ID", axis=1)
```

```python
#check target values
risk_df["Thyroid_Cancer_Risk"].unique()
```

```python
recurrence_df["Recurred"].unique()
```

```python
#Separate Features and Targets
X_risk = risk_df.drop("Thyroid_Cancer_Risk", axis=1)

y_risk = risk_df["Thyroid_Cancer_Risk"]
```

```python
#Encode Target Variables
X_recurrence = recurrence_df.drop("Recurred", axis=1)

y_recurrence = recurrence_df["Recurred"]
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
#Encode TARGET (Risk Dataset)
risk_label_encoder = LabelEncoder()

y_risk_encoded = risk_label_encoder.fit_transform(y_risk)

print(risk_label_encoder.classes_)
print(y_risk_encoded[:10])
```

```python
#Encode TARGET (Recurrence Dataset)
recurrence_label_encoder = LabelEncoder()

y_recurrence_encoded = recurrence_label_encoder.fit_transform(y_recurrence)

print(recurrence_label_encoder.classes_)
print(y_recurrence_encoded[:10])
```

```python
X_risk.dtypes
```

```python
X_recurrence.dtypes
```

```python
#One-hot encoding
X_risk_encoded = pd.get_dummies(X_risk, drop_first=True)
X_recurrence_encoded = pd.get_dummies(X_recurrence, drop_first=True)
```

```python
print(X_risk_encoded.shape)
print(X_recurrence_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
#SPLIT RISK DATASET
X_train_risk, X_test_risk, y_train_risk, y_test_risk = train_test_split(
    X_risk_encoded,
    y_risk_encoded,
    test_size=0.2,
    random_state=42
)
```

```python
#SPLIT RECURRENCE DATASET
X_train_rec, X_test_rec, y_train_rec, y_test_rec = train_test_split(
    X_recurrence_encoded,
    y_recurrence_encoded,
    test_size=0.2,
    random_state=42
)
```

```python
#CHECK SHAPES
print("Risk Train:", X_train_risk.shape)
print("Risk Test:", X_test_risk.shape)

print("Recurrence Train:", X_train_rec.shape)
print("Recurrence Test:", X_test_rec.shape)
```

FEATURE SCALING

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
#Scale Risk Dataset
X_train_risk_scaled = scaler.fit_transform(X_train_risk)
X_test_risk_scaled = scaler.transform(X_test_risk)
```

```python
#scale recurrence dataset
X_train_rec_scaled = scaler.fit_transform(X_train_rec)
X_test_rec_scaled = scaler.transform(X_test_rec)
```

```python
import numpy as np

print(np.mean(X_train_risk_scaled))
print(np.mean(X_train_rec_scaled))
```

```python
#Save processed data and reload it
import numpy as np

np.save("../data/processed/X_train_risk.npy", X_train_risk_scaled)
np.save("../data/processed/X_test_risk.npy", X_test_risk_scaled)
np.save("../data/processed/y_train_risk.npy", y_train_risk)
np.save("../data/processed/y_test_risk.npy", y_test_risk)

np.save("../data/processed/X_train_rec.npy", X_train_rec_scaled)
np.save("../data/processed/X_test_rec.npy", X_test_rec_scaled)
np.save("../data/processed/y_train_rec.npy", y_train_rec)
np.save("../data/processed/y_test_rec.npy", y_test_rec)
```
